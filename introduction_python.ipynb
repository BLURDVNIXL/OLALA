## 2. Variables et types de base

Une **variable** est un nom qu'on donne à une valeur, pour pouvoir la réutiliser plus tard. On crée une variable avec le signe `=` (l'affectation).

````python
prenom = "Alice"
age = 25

print(prenom)
print(age)
````

### Les types de base

Python distingue plusieurs types de données :

| Type    | Nom Python | Exemple        |
|---------|-----------|----------------|
| Entier  | `int`     | `42`           |
| Décimal | `float`   | `3.14`         |
| Texte   | `str`     | `"bonjour"`    |
| Booléen | `bool`    | `True` / `False` |

On peut connaître le type d'une variable avec la fonction `type()`.

````python
entier = 42
decimal = 3.14
texte = "bonjour"
booleen = True

print(type(entier))
print(type(decimal))
print(type(texte))
print(type(booleen))
````

### Conversion de types

On peut convertir une valeur d'un type vers un autre avec `int()`, `float()`, `str()`, `bool()`.

````python
nombre_texte = "10"
nombre = int(nombre_texte)  # convertit le texte "10" en entier 10

print(nombre + 5)  # fonctionne car nombre est bien un entier maintenant

age = 25
age_en_texte = str(age)  # convertit l'entier en texte
print("J'ai " + age_en_texte + " ans")
````

## 3. Opérateurs

### Opérateurs arithmétiques

````python
a = 10
b = 3

print(a + b)   # addition
print(a - b)   # soustraction
print(a * b)   # multiplication
print(a / b)   # division (retourne toujours un float)
print(a // b)  # division entière (arrondie vers le bas)
print(a % b)   # modulo (reste de la division)
print(a ** b)  # puissance (a à la puissance b)
````

### Opérateurs de comparaison

Ils retournent toujours un booléen (`True` ou `False`).

````python
a = 10
b = 3

print(a == b)  # égal à
print(a != b)  # différent de
print(a > b)   # supérieur à
print(a < b)   # inférieur à
print(a >= b)  # supérieur ou égal
print(a <= b)  # inférieur ou égal
````

### Opérateurs logiques

`and`, `or` et `not` permettent de combiner des conditions.

````python
age = 20
a_permis = True

print(age >= 18 and a_permis)   # True seulement si les deux sont vraies
print(age >= 18 or a_permis)    # True si au moins une est vraie
print(not a_permis)             # inverse la valeur
````

## 4. Chaînes de caractères

### Concaténation

On peut assembler des chaînes avec `+`.

````python
prenom = "Marie"
nom = "Curie"
nom_complet = prenom + " " + nom
print(nom_complet)
````

### Les f-strings (la méthode moderne et recommandée)

Les f-strings permettent d'insérer des variables directement dans une chaîne, en préfixant la chaîne par `f` et en mettant les variables entre accolades `{}`.

````python
prenom = "Marie"
age = 55

message = f"{prenom} a {age} ans."
print(message)

# on peut même faire des calculs à l'intérieur
print(f"Dans 10 ans, {prenom} aura {age + 10} ans.")
````

### Méthodes utiles sur les chaînes

````python
texte = "  Bonjour le Monde  "

print(texte.lower())        # met en minuscules
print(texte.upper())        # met en majuscules
print(texte.strip())        # enlève les espaces au début/fin
print(texte.replace("Monde", "Python"))  # remplace
print(len(texte))           # longueur de la chaîne
print(texte.strip().split(" "))  # découpe en liste de mots
````

### Le slicing (découpage)

On peut extraire une portion d'une chaîne en indiquant `[début:fin]`. Attention : en Python, on **compte à partir de 0**, et l'indice de fin est exclu.

````python
mot = "Python"

print(mot[0])      # premier caractère : 'P'
print(mot[0:3])    # du caractère 0 (inclus) au 3 (exclu) : 'Pyt'
print(mot[-1])     # dernier caractère : 'n'
print(mot[len(mot)-1])      # longueur de la chaîne : 6
print(mot[::-1])   # astuce : inverse la chaîne
````

## 5. Structures de données

### Les listes

Une liste permet de stocker **plusieurs valeurs**, dans un ordre précis, entre crochets `[]`.

````python
fruits = ["pomme", "banane", "cerise"]

print(fruits)
print(fruits[0])       # premier élément
print(fruits[-1])      # dernier élément
print(len(fruits))     # nombre d'éléments

fruits.append("mangue")   # ajoute un élément à la fin
print(fruits)

fruits.remove("banane")   # supprime un élément précis
print(fruits)

fruits[0] = "poire"        # modifie un élément
print(fruits)
````

On peut aussi faire du slicing sur les listes, comme pour les chaînes de caractères.

````python
nombres = [10, 20, 30, 40, 50]

print(nombres[1:3])   # du 2e élément (inclus) au 4e (exclu)
print(nombres[:2])    # les 2 premiers
print(nombres[2:])    # à partir du 3e
````

### Les tuples

Un tuple ressemble à une liste, mais il est **immuable** : on ne peut plus le modifier une fois créé. On l'utilise entre parenthèses `()`.

````python
coordonnees = (48.8566, 2.3522)  # latitude, longitude
print(coordonnees)
print(coordonnees[0])

# coordonnees[0] = 0  # provoquerait une erreur : un tuple ne se modifie pas
````

### Les dictionnaires

Un dictionnaire associe des **clés** à des **valeurs**, entre accolades `{}`. C'est très utile pour représenter des données structurées.

````python
personne = {
    "prenom": "Alice",
    "age": 30,
    "ville": "Paris"
}

print(personne)
print(personne["prenom"])   # accéder à une valeur via sa clé

personne["age"] = 31         # modifier une valeur
personne["profession"] = "Ingénieure"  # ajouter une nouvelle clé
print(personne)

print(personne.keys())      # toutes les clés
print(personne.values())    # toutes les valeurs
````

### 🚀 Pour aller plus loin : les sets (ensembles)

Un `set` est une collection d'éléments **uniques** (pas de doublons) et **non ordonnée**. Utile pour éliminer des doublons ou faire des opérations d'ensembles (union, intersection...).

````python
nombres = [1, 2, 2, 3, 3, 3, 4]
uniques = set(nombres)
print(uniques)  # {1, 2, 3, 4}

a = {1, 2, 3}
b = {2, 3, 4}
print(a & b)  # intersection : {2, 3}
print(a | b)  # union : {1, 2, 3, 4}
````

## 6. Structures de contrôle

### Les conditions : `if` / `elif` / `else`

Elles permettent d'exécuter du code seulement si une condition est vraie. **Attention** : en Python, l'indentation (les espaces au début de ligne) est obligatoire et définit les blocs de code.

````python
age = 20

if age < 13:
    print("Tu es un enfant")
elif age < 18:
    print("Tu es un adolescent")
else:
    print("Tu es un adulte")
````

### La boucle `for`

Elle permet de répéter une action pour chaque élément d'une séquence (liste, chaîne, etc.).

````python
fruits = ["pomme", "banane", "cerise"]

for fruit in fruits:
    print(f"J'aime les {fruit}s")
````

La fonction `range()` génère une séquence de nombres, très utile avec les boucles `for`.

````python
for i in range(5):        # de 0 à 4
    print(i)

print("---")

for i in range(2, 10, 2):  # de 2 à 9, par pas de 2
    print(i)
````

### La boucle `while`

Elle répète une action **tant qu'**une condition est vraie.

````python
compteur = 0

while compteur < 5:
    print(f"Compteur = {compteur}")
    compteur = compteur + 1   # important : sinon la boucle ne s'arrête jamais !
````

### `break` et `continue`

- `break` arrête complètement la boucle.
- `continue` passe directement à l'itération suivante.

````python
for i in range(10):
    if i == 5:
        break          # on arrête la boucle dès que i vaut 5
    print(i)

print("---")

for i in range(10):
    if i % 2 == 0:
        continue        # on saute les nombres pairs
    print(i)
````

## 7. Fonctions

Une fonction est un bloc de code réutilisable, auquel on peut donner un nom et des paramètres. Cela évite de répéter le même code plusieurs fois.

### Définir une fonction

````python
def dire_bonjour():
    print("Bonjour !")

dire_bonjour()  # on "appelle" la fonction pour l'exécuter
dire_bonjour()
````

### Fonctions avec paramètres et valeur de retour

Une fonction peut recevoir des **paramètres** (des valeurs en entrée) et **retourner** un résultat avec `return`.

````python
def addition(a, b):
    resultat = a + b
    return resultat

somme = addition(3, 5)
print(somme)

# on peut aussi utiliser directement le résultat
print(addition(10, 20))
````

### Paramètres par défaut

On peut donner une valeur par défaut à un paramètre : elle sera utilisée si on ne fournit pas de valeur lors de l'appel.

````python
def saluer(nom, message="Bonjour"):
    print(f"{message}, {nom} !")

saluer("Alice")               # utilise le message par défaut
saluer("Bob", "Salut")        # remplace le message par défaut
````

### La portée des variables (scope)

Une variable créée **à l'intérieur** d'une fonction n'existe que dans cette fonction : c'est une variable **locale**. Une variable créée en dehors est **globale** et accessible partout.

````python
def exemple():
    variable_locale = "Je n'existe que dans la fonction"
    print(variable_locale)

exemple()

# print(variable_locale)  # provoquerait une erreur : la variable n'existe pas ici
````

## 8. Gestion des erreurs

Parfois, une erreur survient pendant l'exécution du programme (par exemple, diviser par zéro, ou convertir un texte invalide en nombre). On peut "attraper" ces erreurs avec `try` / `except` pour éviter que le programme ne plante.

````python
# Sans gestion d'erreur, ceci ferait planter le programme :
# resultat = 10 / 0

try:
    resultat = 10 / 0
except ZeroDivisionError:
    print("Erreur : division par zéro impossible !")

print("Le programme continue normalement.")
````

````python
def convertir_en_nombre(texte):
    try:
        nombre = int(texte)
        print(f"Conversion réussie : {nombre}")
    except ValueError:
        print(f"Impossible de convertir '{texte}' en nombre.")

convertir_en_nombre("42")
convertir_en_nombre("abc")
````

## 9. Modules

Un module est un fichier Python contenant des fonctions déjà écrites, que l'on peut importer pour les réutiliser. Python fournit de nombreux modules intégrés ("bibliothèque standard").

### Le module `math`

````python
import math

print(math.sqrt(16))     # racine carrée
print(math.pi)            # la constante pi
print(math.floor(4.7))    # arrondi vers le bas
print(math.ceil(4.2))     # arrondi vers le haut
````

### Le module `random`

````python
import random

print(random.randint(1, 10))         # entier aléatoire entre 1 et 10 (inclus)
print(random.choice(["pile", "face"]))  # choix aléatoire dans une liste

nombres = [1, 2, 3, 4, 5]
random.shuffle(nombres)               # mélange la liste sur place
print(nombres)
````

### Importer seulement une partie d'un module

On peut aussi importer uniquement ce dont on a besoin avec `from ... import ...`.

````python
from math import sqrt, pi

print(sqrt(25))
print(pi)
````

### Les compréhensions de liste

Elles permettent de créer une liste en une seule ligne, à partir d'une boucle et éventuellement d'une condition.

````python
# Méthode classique
carres = []
for i in range(10):
    carres.append(i ** 2)
print(carres)

# Avec une compréhension de liste (équivalent, plus concis)
carres = [i ** 2 for i in range(10)]
print(carres)

# Avec une condition
pairs = [i for i in range(20) if i % 2 == 0]
print(pairs)
````

### Lecture et écriture de fichiers

Python permet de lire et écrire des fichiers texte avec `open()`. Il est recommandé d'utiliser `with`, qui ferme automatiquement le fichier après usage.

````python
# Écriture dans un fichier
with open("exemple.txt", "w") as fichier:
    fichier.write("Bonjour depuis Python !\n")
    fichier.write("Deuxième ligne.")

# Lecture du fichier
with open("exemple.txt", "r") as fichier:
    contenu = fichier.read()
    print(contenu)
````
